```r
#Set plot dimensions
options(repr.plot.width = 12, repr.plot.height = 7) 

#Count team appearances in each tournament finish category
tourney_finish_proportions <- women_march_tidy |>
group_by(tourney_finish) |> summarize(count = n()) |>
  
#Calculate proportion of each tournament finish category
mutate(proportion = count / sum(count))

#Create a bar plot shows the proportion in each tournament-finish category
tourney_finish_plot <- ggplot(tourney_finish_proportions,aes(x = fct_reorder(tourney_finish, proportion, .desc = TRUE), y = proportion)) +
geom_bar(stat = "identity", fill = "steelblue", color = "white") +
labs(x = "Stage Reached in the Tournament",y = "Proportion of Team Appearances") +
ggtitle("Proportions of Tournament Finishes") + theme(text = element_text(size=15))
tourney_finish_plot
```

Visualization 1: The scatterplot shows the relationsip between tournament seed and regional wins with colours showing stage reached in the tournament. It helps us check whether seed and regional wins are useful to predict tournament finish. In this plot, many colours overlap which indicate team with similar seeds and number of region wins can have different tournament finishes. This may make the response variable hard to predict using only these two predictor variables.

```r
#Set plot dimensions
options(repr.plot.width = 12, repr.plot.height = 7)

#Create scatterplot shows relationship between two predictors
relationship_plot <- women_march_tidy |> mutate(tourney_finish = fct_recode(tourney_finish, 
"First Round Loss" = "first_round_loss",
"Second Round Loss" = "second_round_loss",
"Top 16 Loss" = "top_16_loss",
"Top Eight Loss" = "top_8_loss",
"Final Four Loss" = "top_4_loss",
"Championship Loss" = "top_2_loss",
"Champion" = "champ")) |>
ggplot( aes(x = seed, y = reg_wins, color = tourney_finish)) +
geom_point(alpha = 0.6) +
labs(x = "Tournament Seed",y = "Number of Regional Wins",color = "Stage Reached in the Tournament")+
ggtitle("Stage Reached in the Tournament \n by Seed and Regional Wins")+
theme(text = element_text(size = 20))
relationship_plot
```

Visualization 2: The bar plot shows the average number of regional wins for each stage reached in the tournament. It helps us compare and see whether teams can advance further tend to have more regional wins. If the averages are different, number of regional wins may help us predict the tournament finish. Note that each bar shows a group average, not the results of every team.

```r
#Calculate average regional wins by each tournament finish categoriy
tourney_finish_avg_wins <- women_march_tidy |>group_by(tourney_finish) |>
summarize(avg_reg_wins = mean(reg_wins))

#Set plot dimensions
options(repr.plot.width = 10, repr.plot.height = 6)

#Create bar plot shows average of regional wins for each tournament finish category
tourney_finish_plot <- tourney_finish_avg_wins |> mutate(tourney_finish = fct_recode(tourney_finish, 
"First Round \n Loss" = "first_round_loss",
"Second Round \n Loss" = "second_round_loss",
"Top 16 \n Loss" = "top_16_loss",
"Top Eight \n Loss" = "top_8_loss",
"Final Four \n Loss" = "top_4_loss",
"Championship \n Loss" = "top_2_loss",
"Champion" = "champ")) |>
ggplot(aes(x = tourney_finish, y = avg_reg_wins)) +
geom_bar(stat = "identity", fill = "steelblue", color = "white", width = 0.7) +
labs(x = "Stage Reached in the Tournament", y = "Average Regional Wins")+
ggtitle("Average Regional Wins by Tournament Finish") +
theme(text = element_text(size = 18))
tourney_finish_plot
```
